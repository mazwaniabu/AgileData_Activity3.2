# CI/CD Duplicate Removal Notebook
This notebook demonstrates a simple duplicate removal pipeline using pandas.

## Agile Data Science
Agile Data Science focuses on short feedback loops, iterative delivery, and automated validation.
In this activity, we validate each change using tests and CI workflows.

```python
import pandas as pd
from pathlib import Path
```

## Load Dataset
We load the raw dataset from the data/raw folder.

```python
input_path = Path("data/raw/dataset.csv")
df = pd.read_csv(input_path)
df.head()
```

## Detect Duplicates
We check how many duplicate rows exist.

```python
duplicate_count = df.duplicated().sum()
duplicate_count
```

## Remove Duplicates
We drop duplicate rows and reset the index.

```python
cleaned = df.drop_duplicates().reset_index(drop=True)
cleaned.head()
```

## Save Cleaned Dataset
We save the cleaned dataset to the data/process folder.

```python
output_path = Path("data/process/dataset.csv")
cleaned.to_csv(output_path, index=False)
output_path.exists()
```

## Before and After Comparison
We compare row counts before and after cleaning.

```python
before_rows = len(df)
after_rows = len(cleaned)
before_rows, after_rows
```

## Conclusion
We successfully detected and removed duplicate rows, and saved the cleaned dataset.
